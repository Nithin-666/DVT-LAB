```python
# ============================================================
# EXPERIMENT 4
# DATA EXPLORATION AND DESCRIPTIVE STATISTICS
# WORLD HAPPINESS DATASET 2015-2024
# ============================================================


# ============================================================
# 1. IMPORT DATASET DIRECTLY FROM KAGGLE
# ============================================================

!pip install -q kagglehub

import kagglehub
import pandas as pd
import glob
import os


# Download dataset from Kaggle
path = kagglehub.dataset_download(
    "yadiraespinoza/world-happiness-2015-2024"
)

print("Dataset downloaded successfully!")
print("Path:", path)


# ============================================================
# 2. FIND CSV FILES
# ============================================================

csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

print("\nCSV files found:")

for file in csv_files:
    print(os.path.basename(file))


# ============================================================
# 3. READ AND COMBINE ALL CSV FILES
# ============================================================

dataframes = []

for file in csv_files:

    temp = pd.read_csv(file, delimiter=';') # Added delimiter parameter

    # Keep the year/file information
    temp["Year_File"] = os.path.basename(file)

    dataframes.append(temp)


df = pd.concat(
    dataframes,
    ignore_index=True
)


# Display first 5 rows
print("\nFirst 5 rows:")
print(df.head())


# ============================================================
# 3.1. Convert numeric columns with comma decimals to float
# ============================================================

for col in df.columns:
    # Check if the column is of object type (likely strings)
    if df[col].dtype == 'object':
        # Check if any value in the column contains a comma
        if df[col].astype(str).str.contains(',').any():
            try:
                # Replace comma with dot and convert to numeric
                df[col] = df[col].astype(str).str.replace(',', '.', regex=False).astype(float)
            except ValueError:
                # If conversion fails, it's not a numeric column with comma decimals
                pass


# ============================================================
# 4. DATASET SHAPE
# ============================================================

print("\nDataset Shape:")

print(df.shape)


# ============================================================
# 5. DATA TYPES
# ============================================================

print("\nData Types:")

print(df.dtypes)


# ============================================================
# 6. FIND HAPPINESS SCORE COLUMN
# ============================================================

if "Life Ladder" in df.columns:

    score_column = "Life Ladder"

elif "Ladder score" in df.columns:

    score_column = "Ladder score"

elif "Happiness Score" in df.columns:

    score_column = "Happiness Score"

else:

    raise ValueError(
        "Happiness score column not found."
    )


# ============================================================
# 7. RECENT HAPPINESS RECORDS
# ============================================================

# The World Happiness dataset covers multiple years.
# Select records with a happiness score above the median.

median_score = df[score_column].median()

high_happiness = df[
    df[score_column] > median_score
]

print("\nRecords with happiness score above median:")

print(high_happiness)


# ============================================================
# 8. SELECT A COUNTRY
# ============================================================

if "Country name" in df.columns:

    country_column = "Country name"

elif "Country" in df.columns:

    country_column = "Country"

else:

    raise ValueError(
        "Country column not found."
    )


# Select United States records
us_titles = df[
    df[country_column] == "United States"
]

print("\nUnited States Happiness Records:")

print(us_titles)


# ============================================================
# 9. SORT DATA
# ============================================================

# Sort by happiness score

sorted_df = df.sort_values(
    by=score_column,
    ascending=False
)

print("\nDataset Sorted by Happiness Score:")

print(sorted_df)


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics for Happiness Score")

print(
    "Mean =",
    df[score_column].mean()
)

print(
    "Median =",
    df[score_column].median()
)

print(
    "Mode =",
    df[score_column].mode()[0]
)

print(
    "Standard Deviation =",
    df[score_column].std()
)


# ============================================================
# 11. DESCRIBE DATASET
# ============================================================

print("\nDataset Description:")

print(df.describe())


# ============================================================
# 12. GROUPBY ANALYSIS
# ============================================================

# Average happiness score by year/file

year_avg = df.groupby(
    "Year_File"
)[score_column].mean()

print("\nAverage Happiness Score by Year:")

print(year_avg)


# ============================================================
# 13. AVERAGE HAPPINESS BY COUNTRY
# ============================================================

country_avg = df.groupby(
    country_column
)[score_column].mean()

print("\nAverage Happiness Score by Country:")

print(country_avg)
```
